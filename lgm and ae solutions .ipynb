```python
 #logistic growth model  
import numpy as np
from scipy.integrate import odeint
import matplotlib.pyplot as plt
def module(y,t,r):
    dydt = y*r*(1-y/k)
    return dydt
y0=1000
k=500
t= np.linspace(0,10)
r=0.2
y1 = odeint(module,y0,t, args =(r,))
r=0.5
y2 = odeint(module,y0,t, args =(r,))
r=1.0
y3 = odeint(module,y0,t, args =(r,))
plt.plot(t,y1,'r',linewidth=2,label= 'r=0.2')
plt.plot(t,y2,'g',linewidth=2,label= 'r=0.5')
plt.plot(t,y3,'b',linewidth=2,label= 'r=1.0')
plt.xlabel('t')
plt.ylabel('N')
plt.legend()
plt.show()
```

```python
#exponential growth model
import numpy as np
from scipy.integrate import odeint
import matplotlib.pyplot as plt
def module(y,t,r):
    dydt = y*r
    return dydt
y0=100
t= np.linspace(0,50)
r=0.10
y1 = odeint(module,y0,t, args =(r,))
r=0.11
y2 = odeint(module,y0,t, args =(r,))
r=0.12
y3 = odeint(module,y0,t, args =(r,))
plt.plot(t,y1,'r',linewidth=2,label= 'r=0.10')
plt.plot(t,y2,'g',linewidth=2,label= 'r=0.11')
plt.plot(t,y3,'b',linewidth=2,label= 'r=0.12')
plt.xlabel('t')
plt.ylabel('N')
plt.legend()
plt.show()
```

```python
#logistic growth model with alee effect
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
def module(y,t,m):
    dydt = y*(1-y/k)*(y/m -1)
    return dydt
k=500
y0=100
t=np.linspace(0,20)
m =50
y1 = odeint(module,y0,t, args =(m,))
m=150
y2 = odeint(module,y0,t, args =(m,))
m=100
y3 = odeint(module,y0,t, args =(m,))

plt.plot(t,y1,'r',linewidth=2, label=' M=50(M< initial pop.)')
plt.plot(t,y2,'g',linewidth=2, label='M=150(M> initial pop.)')
plt.plot(t,y3,'b',linewidth=2, label=' M=100(M= initial pop.)')
plt.xlabel('t')
plt.ylabel('N')
plt.legend()
plt.show()
```

```python
#bifurcation diagram in lgm due to constant harversting
import numpy as np
from scipy.integrate import odeint
import matplotlib.pyplot as plt
def module(y,t,H):
    dydt = y*r*(1-y/k) - H
    return dydt
y0=5.0
k=10
r=0.2
t= np.linspace(0,30)
H=0.10
y1 = odeint(module,y0,t, args =(H,))
H=0.50
y2 = odeint(module,y0,t, args =(H,))
H=0.60
y3 = odeint(module,y0,t, args =(H,))
plt.plot(t,y1,'blue',linewidth=2,label= 'H<Kr/4')
plt.plot(t,y2,'seagreen',linewidth=2,label= 'H=Kr/4')
plt.plot(t,y3,'r',linewidth=2,label= 'H>Kr/4')
plt.xlabel('t')
plt.ylabel('N')

plt.legend()
plt.show()
```
