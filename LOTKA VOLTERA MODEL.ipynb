```python
#E(0,0)
import numpy as np
import matplotlib.pyplot as plt 
from scipy.integrate import odeint

def dNdx(x,N):
 y1,y2=N
 a=1
 b=1
 return [y1*(1-y1)-a*y1*y2,
 y2*(1-y2)-b*y1*y2]
y1_0=0
y2_0=0
N_0=(y1_0,y2_0)
x=np.linspace(0,50,100)
sol = odeint( dNdx, y0=N_0,t=x, tfirst=True) 
y1_sol=sol.T[0]
y2_sol=sol.T[1]
plt.plot(x,y1_sol,'g')
plt.plot(x,y2_sol,'b')
plt.show()

```

```python
#E(A*,B*) CASE
import numpy as np
import matplotlib.pyplot as plt 
from scipy.integrate import odeint

def dNdx(x,N):
 y1,y2=N
 a=1.0
 b=2.0
 return [y1*(1-y1)-a*y1*y2,
 y2*(1-y2)-b*y1*y2]
y1_0=(1-a)/(1-a*b)
y2_0=(1-b)/(1-a*b)
N_0=(y1_0,y2_0)
x=np.linspace(0,50,100)
sol = odeint( dNdx, y0=N_0,t=x, tfirst=True) 
y1_sol=sol.T[0]
y2_sol=sol.T[1]
plt.plot(x,y1_sol,'g')
plt.plot(x,y2_sol,'b')
plt.show()
```
