```python
# import libraries
import numpy as np
import matplotlib.pyplot as plt


def logistic(r, x):
    return r * x * (1 - x)
n = 100000  # total number of points
r = np.linspace(0.0, 4.0, n)
iterations = 100000

x = np.random.uniform(0.0, 1.0, n)  # initial value
for _ in range(iterations):  # iterate
    x = logistic(r, x)
fig = plt.figure(figsize=(15, 10))
ax = plt.axes()

ax.set_title(f"Bifurcation Diagram ", fontsize=24)
ax.set_xlabel('r', fontsize=24)
ax.set_ylabel('x', fontsize=24)
ax.tick_params(axis='both', which='major', labelsize=18)
ax.set_xlim(3.999, 4.0)
ax.set_ylim(0.0, 1.0)


ax.plot(r, x, ',r', alpha=1.0)
plt.show()
```
