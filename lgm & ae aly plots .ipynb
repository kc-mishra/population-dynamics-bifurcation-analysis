```python
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(0,50,100)
r=0.2
k=50
y1= r*x*(1-x/k)
r=0.3
y2= r*x*(1-x/k)
r=0.4
y3= r*x*(1-x/k)
plt.plot(x,y1,'r',linewidth=2,label= 'r=0.2')
plt.plot(x,y2,'b',linewidth=2,label= 'r=0.3')
plt.plot(x,y3,'y',linewidth=2,label= 'r=0.4')
plt.xlabel('N')
plt.ylabel('dN/dt')

plt.legend()
plt.show()
```
