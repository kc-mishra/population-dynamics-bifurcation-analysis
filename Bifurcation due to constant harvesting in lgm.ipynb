```python
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(0,50,100)
r=0.2
k=50
h=1.0
y1= r*x*(1-x/k)-h
h=2.5
y2= r*x*(1-x/k)-h
h=4
y3= r*x*(1-x/k)-h
y4= x*0
plt.plot(x,y1,'r',label='H<Kr/4')
plt.plot(x,y2,'b',label='H=Kr/4')
plt.plot(x,y3,'y',label='H>Kr/4')
plt.plot(x,y4,'black', linewidth=0.5)
plt.xlabel('N')
plt.ylabel('dN/dt')
plt.title('LGM with Bifurcations due to harvesting(H)')
plt.legend()

plt.show()
```

```python
#bifurcation diagram in lgm
import numpy as np
from scipy.integrate import odeint
import matplotlib.pyplot as plt
def module(y,t,H):
    dydt = y*r*(1-y/k) - H
    return dydt
y0=5.0
k=10
r=0.2
t= np.linspace(0,30)
H=0.10
y1 = odeint(module,y0,t, args =(H,))
H=0.50
y2 = odeint(module,y0,t, args =(H,))
H=0.60
y3 = odeint(module,y0,t, args =(H,))
plt.plot(t,y1,'blue',linewidth=2,label= 'H<Kr/4')
plt.plot(t,y2,'seagreen',linewidth=2,label= 'H=Kr/4')
plt.plot(t,y3,'r',linewidth=2,label= 'H>Kr/4')
plt.xlabel('t')
plt.ylabel('N')

plt.legend()
plt.show()
```
